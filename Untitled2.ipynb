```python
import numpy as np
import matplotlib.pyplot as plt

print("Board Game Probability Project")
print("Python is ready!")
```

```python
# We use 10 positions: 0 to 9
N = 10

# A ladder moves the player forward
# A snake moves the player backward
jumps = {
    2: 6,   # ladder
    5: 1,   # snake
    7: 9    # ladder
}

print("Number of states:", N)
print("Special positions:", jumps)
```

```python
P = np.zeros((N, N))

# A fair dice has 6 possible results
for state in range(N):

    # State 9 means the player has already won.
    # Therefore, they stay at state 9.
    if state == 9:
        P[state, state] = 1
        continue

    for dice in range(1, 7):

        next_state = state + dice

        # If the dice takes us beyond 9,
        # we remain at 9.
        next_state = min(next_state, 9)

        # Apply snake or ladder
        if next_state in jumps:
            next_state = jumps[next_state]

        P[state, next_state] += 1 / 6

print("Transition Matrix:")
print(np.round(P, 3))

print("\nEach row sum:")
print(P.sum(axis=1))
```

```python
turns = int(input("Enter number of turns: "))

# Initially, the player is at state 0
initial = np.zeros(N)
initial[0] = 1

# Probability distribution after the chosen number of turns
distribution = initial @ np.linalg.matrix_power(P, turns)

print("\nProbability of being in each state:")
for state, probability in enumerate(distribution):
    print(f"State {state}: {probability:.4f}")

print("\nProbability of winning:", distribution[9])
```

```python
# Number of turns
turns = 10

# Initially, the player is at state 0
initial = np.zeros(N)
initial[0] = 1

# Calculate probability distribution after 10 turns
distribution = initial @ np.linalg.matrix_power(P, turns)

print("Probability after", turns, "turns:")

for state, probability in enumerate(distribution):
    print(f"State {state}: {probability:.4f}")

print("\nProbability of winning:", distribution[9])

print("\nTotal probability:", distribution.sum())
```

```python
states = np.arange(N)

plt.figure(figsize=(10, 5))

plt.bar(states, distribution)

plt.xlabel("Board State")
plt.ylabel("Probability")
plt.title(f"Probability Distribution After {turns} Turns")

plt.xticks(states)
plt.ylim(0, 1)

plt.show()
```

```python
long_turns = 100

long_distribution = initial @ np.linalg.matrix_power(P, long_turns)

print("Probability distribution after", long_turns, "turns:")

for state, probability in enumerate(long_distribution):
    print(f"State {state}: {probability:.6f}")

print("\nLong-run probability of being at the winning state:")
print(long_distribution[9])
```

```python
eigenvalues, eigenvectors = np.linalg.eig(P)

print("Eigenvalues:")
for value in eigenvalues:
    print(round(value.real, 4))
```
